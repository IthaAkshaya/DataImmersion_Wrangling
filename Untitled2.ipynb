```python
import pandas as pd
from google.colab import files

# STEP 1: Upload your CSV file into Colab
# Run this cell, then choose your local sales_data.csv file
uploaded = files.upload()

# STEP 2: Load dataset
df_raw = pd.read_csv("sales_data.csv", sep="|", index_col=False)
df_raw = df_raw.dropna(how='all')
df_raw = df_raw.loc[:, ~df_raw.columns.str.contains('^Unnamed')]
df_raw.columns = df_raw.columns.str.strip()
df_raw.columns = df_raw.columns.str.replace(r"\s+", "", regex=True)
df_raw.columns = df_raw.columns.str.replace(r"\|", "", regex=True)

print("=== TASK 1: RAW DATASET (Before Cleaning) ===")
print(df_raw)

data_dict_raw = pd.DataFrame({
    "Column": df_raw.columns,
    "DataType": [str(df_raw[col].dtype) for col in df_raw.columns],
    "NonNullCount": [df_raw[col].notnull().sum() for col in df_raw.columns],
    "UniqueValues": [df_raw[col].nunique() for col in df_raw.columns]
})
print("\n=== RAW DATA DICTIONARY ===")
print(data_dict_raw)
print("Total rows (raw):", len(df_raw))

# STEP 3: Cleaning
df_clean = df_raw.drop_duplicates()
if 'Gender' in df_clean.columns:
    df_clean['Gender'] = df_clean['Gender'].replace({'M':'Male','F':'Female'})
if 'Quantity' in df_clean.columns:
    df_clean['Quantity'] = pd.to_numeric(df_clean['Quantity'], errors='coerce')
if 'Price' in df_clean.columns:
    df_clean['Price'] = pd.to_numeric(df_clean['Price'], errors='coerce')
if 'DOB' in df_clean.columns:
    df_clean['DOB'] = pd.to_datetime(df_clean['DOB'], errors='coerce')
if 'PurchaseDate' in df_clean.columns:
    df_clean['PurchaseDate'] = pd.to_datetime(df_clean['PurchaseDate'], errors='coerce', dayfirst=True)
if 'Price' in df_clean.columns:
    df_clean = df_clean[df_clean['Price'] >= 0]
if 'DOB' in df_clean.columns:
    df_clean['Age'] = (pd.Timestamp("2024-01-01") - df_clean['DOB']).dt.days // 365

print("\n=== TASK 2: CLEANED DATASET (After Cleaning) ===")
print(df_clean)

data_dict_clean = pd.DataFrame({
    "Column": df_clean.columns,
    "DataType": [str(df_clean[col].dtype) for col in df_clean.columns],
    "NonNullCount": [df_clean[col].notnull().sum() for col in df_clean.columns],
    "UniqueValues": [df_clean[col].nunique() for col in df_clean.columns]
})
print("\n=== CLEANED DATA DICTIONARY ===")
print(data_dict_clean)
print("Total rows (cleaned):", len(df_clean))

# STEP 4: Export both raw and cleaned datasets + dictionaries
df_raw.to_excel("raw_sales_data.xlsx", index=False)
df_clean.to_excel("cleaned_sales_data.xlsx", index=False)
data_dict_raw.to_excel("raw_data_dictionary.xlsx", index=False)
data_dict_clean.to_excel("cleaned_data_dictionary.xlsx", index=False)

print("\nExcel files exported successfully.")

# STEP 5: Download all files for GitHub upload
files.download("raw_sales_data.xlsx")
files.download("cleaned_sales_data.xlsx")
files.download("raw_data_dictionary.xlsx")
files.download("cleaned_data_dictionary.xlsx")

# STEP 6: Save and download the Python script itself
with open("task1_cleaning.py", "w") as f:
    f.write("""<paste this entire code here>""")

files.download("task1_cleaning.py")
```
